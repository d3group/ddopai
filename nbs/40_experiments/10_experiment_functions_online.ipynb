````python
#| export

class EarlyStoppingHandler():

    '''
    Class to handle early stopping during experiments. The EarlyStoppingHandler handler calculates the average
    score over the last "patience" epochs and compares it to the average score over the previous "patience" epochs.
    Note that one epoch we define here as time in between evaluating on a validation set, for supervised learning
    typically one epoch is one pass through the training data. For reinforcement learning, in between each evaluation
    epoch there may be less than one, one, or many episodes played in the training environment.

    '''
    def __init__(
        self,
        patience: int = 50, # Number of epochs to evaluate for stopping
        warmup: int = 100, # How many initial epochs to wait before evaluating
        criteria: str = "J",  # Whether to use discounted rewards J or total rewards R as criteria
        direction: str = "max"  # Whether reward shall be maximized or minimized
    ):

        self.history = list()
        self.patience = patience
        if warmup is None or warmup < patience * 2:
            warmup = patience * 2
        self.warmup = warmup
        self.criteria = criteria
        self.direction = direction

    def add_result(self,
                    J: float, # Return (discounted rewards) of the last epoch
                    R: float, # Total rewards of the last epoch
                    ) -> bool:

        """
        Add the result of the last epoch to the history and check if the experiment should be stopped.

        """
        if self.criteria == "J":
            self.history.append(J)
        elif self.criteria == "R":
            self.history.append(R)
        else:
            raise ValueError("Criteria must be J or R")
        
        if len(self.history) >= self.warmup:
            if self.direction == "max":
                if sum(self.history[-self.patience*2:-self.patience]) >= sum(self.history[-self.patience:]):
                    return True
                else:
                    return False
            elif self.direction == "min":
                if sum(self.history[-self.patience*2:-self.patience]) <= sum(self.history[-self.patience:]):
                    return True
                else:
                    return False
            else:
                raise ValueError("Direction must be max or min")
````

````python
#| export

class DatasetCallback:
    def __init__(self):
        self.dataset = []

    def __call__(self, dataset):
        self.dataset.append(dataset)
    
    def reset(self):
        self.dataset = []
    
    def get_dataset(self):
        return self.dataset
        
def calculate_score(
                    dataset: List,
                    env: BaseEnvironment, # Any environment inheriting from BaseEnvironment
                    ) -> Tuple[float, float]:

    """

    Calculate the total rewards R and the discounted rewards J of a dataset.

    """

    R = sum([row[0][2] for row in dataset])
    gamma = env.mdp_info.gamma
    J = sum([gamma**(t) * row[0][2] for t, row in enumerate(dataset)]) # Note: t starts at 1 so the first reward is already discounted

    return R, J

def log_info(R: float,
                J: float,
                n_epochs: int,
                tracking: Literal["wandb"], # only wandb implemented so far
                mode: Literal["train", "val", "test"]
                ):
    
    '''
    Logs the same R, J information repeatedly for n_epoochs.
    E.g., to draw a straight line in wandb for algorithmes
    such as XGB, RF, etc. that can be comparared to the learning
    curves of supervised or reinforcement learning algorithms.
    '''

    if tracking == "wandb":
        for epoch in range(n_epochs):
            wandb.log({f"{mode}/R": R, f"{mode}/J": J})
    else:
        pass

def update_best(R: float, J: float, best_R: float, best_J: float): # 
    
    """

    Update the best total rewards R and the best discounted rewards J.

    """

    if R > best_R:
        best_R = R
    if J > best_J:
        best_J = J

    return best_R, best_J

def save_agent(agent: BaseAgent, # Any agent inheriting from BaseAgent
                experiment_dir: str, # Directory to save the agent, 
                save_best: bool,
                R: float,
                J: float,
                best_R: float,
                best_J: float,
                criteria: str = "J",
                force_save = False,
                ):

    """
    Save the agent if it has improved either R or J, depending on the criteria argument,
    vs. the previous epochs

    """

    if save_best:
        if criteria == "R":
            if R == best_R:
                save_dir = f"{experiment_dir}/saved_models/best"
                agent.save(save_dir)
            elif force_save:
                save_dir = f"{experiment_dir}/saved_models/best"
                agent.save(save_dir)
        elif criteria == "J":
            if J == best_J:
                save_dir = f"{experiment_dir}/saved_models/best"
                agent.save(save_dir)
            elif force_save:
                save_dir = f"{experiment_dir}/saved_models/best"
                agent.save(save_dir)
````

````python
#| export

def evaluate_val(*,
                 agent,
                 envs: List,
                 epoch: int,
                 core_cls: Core,
                 use_wandb: bool = False,
                 eval_step_info: bool = False):
    """
    Evaluate **validation** performance on a list of envs.
    Every env is switched to `.val()` mode, run for ONE full episode,
    then switched back to `.train()`.

    Parameters
    ----------
    agent : BaseAgent
    envs  : list[BaseEnvironment]   – validation envs
    epoch : int                     – originating training epoch
    log_to_wandb : bool
    eval_step_info : bool
    """
    logging.info(f"[VAL] running validation after epoch {epoch+1}")

    # ensure no training ops during evaluation
    agent.eval()

    all_R, all_J = [], []
    all_Cum_True_R, all_Cum_R = [], []
    for idx, venv in enumerate(envs):
        venv.reset()  # reset to initial state
        #agent.reset()  # reset agent to initial state
        # -------- build core with callback that records ds for score ----
        core = core_cls(agent, venv)

        # update env reference inside the agent if it needs it (e.g., changing price grids)
        if epoch > 0 and hasattr(agent, "update_env"):
            agent.update_env(venv)
        if hasattr(agent, "reset_hidden"):
            agent.reset_hidden()
        # -------------------------------------------------- training step
        dataset = core.evaluate(quiet=True,
                   n_episodes=1)
        
        info_hist = venv.get_info_history()
        R, J = calculate_score(dataset, venv)
        all_R.append(R); all_J.append(J)

        if use_wandb:
            cumulative_reward, cumulative_true_reward = 0.0, 0.0
            for t, info in info_hist.items():
                cumulative_reward += np.squeeze(info["reward"])
                cumulative_true_reward += np.squeeze(info.get("true_reward", info["reward"]))
                if "inv" in info:
                    wandb.log({"Inventory": np.squeeze(info["inv"])}, commit=False)
                wandb.log(venv.env_type, commit=False)
                wandb.log({
                    "Epoch": epoch,
                    "Val_Epoch": idx,
                    "Val_t": t,
                    "Val_Action": info["action"],
                    "Val_Reward": np.squeeze(info["reward"]),
                    "Val_True_Reward": np.squeeze(info.get("true_reward", info["reward"])),
                    "Val_Cumulative_Reward": cumulative_reward,
                    "Val_Cumulative_True_Reward": cumulative_true_reward,
                    "Val_R": R,
                    "Val_J": J,
                }, commit=True)
            all_Cum_R.append(cumulative_reward)
            all_Cum_True_R.append(cumulative_true_reward)

        venv.train()          # restore original mode

    logging.info(f"[VAL] mean True_R={np.mean(all_Cum_True_R):.2f}  mean R={np.mean(all_Cum_R):.2f}")
    del(info_hist)
    del(dataset)
    # switch agent back
    agent.train()
    return np.mean(all_Cum_R), np.mean(all_Cum_True_R)

# =================================================================
# 2) test‑set evaluation  (final / only‑eval)
# =================================================================
def evaluate_test(*,
                  agent,
                 envs: List,
                 epoch: int,
                 core_cls: Core,
                 use_wandb: bool = False,
                 eval_step_info: bool = False):
    """
    Evaluate **test** performance (no epoch reference).
    Each env is put into `.test()` mode if available, otherwise `.val()`.
    """
    logging.info("[TEST] running final evaluation")

    agent.eval()

    all_R, all_J = [], []
    all_Cum_True_R, all_Cum_R = [], []
    for idx, tenv in enumerate(envs):
        # prefer explicit .test() method if env implements it
        
        tenv.reset()  # reset to initial state
        #agent.reset()  # reset agent to initial state
        # -------- build core with callback that records ds for score ----
        core = core_cls(agent, tenv)

        # update env reference inside the agent if it needs it (e.g., changing price grids)
        if epoch > 0 and hasattr(agent, "update_env"):
            agent.update_env(tenv)
        if hasattr(agent, "reset_hidden"):
            agent.reset_hidden()
        dataset = core.evaluate(quiet=True,
                   n_episodes=1)
        
        info_hist = tenv.get_info_history()
        R, J = calculate_score(dataset, tenv)
        all_R.append(R); all_J.append(J)

        if use_wandb:
            cumulative_reward, cumulative_true_reward = 0.0, 0.0
            for t, info in info_hist.items():
                cumulative_reward += np.squeeze(info["reward"])
                cumulative_true_reward += np.squeeze(info.get("true_reward", info["reward"]))
                if "inv" in info:
                    wandb.log({"Inventory": np.squeeze(info["inv"])}, commit=False)
                wandb.log({
                    "Epoch": epoch,
                    "Test_Epoch": idx,
                    "Test_t": t,
                    "Test_Action": info["action"],
                    "Test_Reward": np.squeeze(info["reward"]),
                    "Test_True_Reward": np.squeeze(info.get("true_reward", info["reward"])),
                    "Test_Cumulative_Reward": cumulative_reward,
                    "Test_Cumulative_True_Reward": cumulative_true_reward,
                    "Test_R": R,
                    "Test_J": J,
                }, commit=True)
            all_Cum_R.append(cumulative_reward)
            all_Cum_True_R.append(cumulative_true_reward)
        # switch back to train to avoid side effects when this function is
        # invoked before more training (rare but safe)
        tenv.train()
    del(info_hist)
    del(dataset)
    logging.info(f"[VAL] mean True_R={np.mean(all_Cum_True_R):.2f}  mean R={np.mean(all_Cum_R):.2f}")

    agent.train()
    return np.mean(all_Cum_R), np.mean(all_Cum_True_R)
````

````python
#| export
def run_experiment(
        *,
        agent,                                     # BaseAgent
        train_envs:   List = None,                 # envs used for training
        eval_envs:    List = None,                 # envs used for val / test
        experiment_type: str = "train_only",       # 'train_only'|'train_val'|'only_eval'
        # ---------- training hyper‑params ----------
        n_epochs: int = 1,
        n_steps: int = None,
        n_steps_per_fit: int = 1,
        n_episodes_per_fit: int = None,
        # ---------- validation ----------
        val_every: int = 10,                       # how often to call evaluate_val
        # ---------- misc ----------
        early_stopping_handler: Union[EarlyStoppingHandler, None] = None,
        save_best: bool = True,
        performance_criterion: str = "J",
        tracking: Union[str, None] = None,         # 'wandb' or None
        results_dir: str = "results",
        run_id: str | None = None,
        print_freq: int = 1,
        eval_step_info: bool = False,
):
    """
    Unifies three modes:
      • train_only   – replicate old behaviour (no validation)
      • train_val    – train & periodically call evaluate_val
      • only_eval    – skip training & immediately call evaluate_test
    """

    assert experiment_type in {"train_only", "train_val", "only_eval"}, \
        f"unknown experiment_type '{experiment_type}'"

    # ------------------------------------------------------------------
    # create experiment folder / run id
    # ------------------------------------------------------------------
    if run_id is None:
        run_id = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
    experiment_dir = f"{results_dir}/{run_id}"
    logging.info(f"★ Experiment dir: {experiment_dir}")

    # Convenience handles
    use_wandb = tracking == "wandb"
    core_cls  = MetaCore if getattr(agent, "agent_name", "") == "RL2PPO" else Core

    # ==================================================================
    # 1) ONLY‑EVAL MODE
    # ==================================================================
    if experiment_type == "only_eval":
        logging.info("Running *evaluation‑only* – no training.")
        assert eval_envs is not None, "`eval_envs` required for only_eval"
        evaluate_test(agent=agent,
                      envs=eval_envs,
                      core_cls=core_cls,
                      use_wandb=use_wandb,
                      eval_step_info=eval_step_info)
        return
    # everything below is training‑related
    assert train_envs is not None, "`train_envs` must be provided for training"

    # ==================================================================
    # initialise bookkeeping
    # ==================================================================
    best_R, best_J = -np.inf, -np.inf

    # ==================================================================
    # 2) TRAINING LOOP  (train_only  OR  train_val)
    # ==================================================================
    for epoch in trange(n_epochs, desc="epochs"):
        #env = train_envs[epoch % len(train_envs)]
        env = np.random.choice(train_envs)
        env.reset()

        # -------- build core with callback that records ds for score ----
        core = core_cls(agent, env, callbacks_fit=[])

        # update env reference inside the agent if it needs it (e.g., changing price grids)
        if epoch > 0 and hasattr(agent, "update_env"):
            agent.update_env(env)

        # -------------------------------------------------- training step
        core.learn(n_steps=n_steps,
                   n_steps_per_fit=n_steps_per_fit,
                   n_episodes_per_fit=n_episodes_per_fit,
                   quiet=True,
                   n_episodes=1)

        # -------------------------------------------------- bookkeeping
        info_hist = env.get_info_history()

        # -------------------------------------------------- wandb train log
        if use_wandb:
            cumulative_reward = 0.0
            cumulative_true_reward = 0.0
            for t, info in info_hist.items():
                cumulative_reward += np.squeeze(info["reward"])
                cumulative_true_reward += np.squeeze(info.get("true_reward", info["reward"]))
                if "inv" in info:
                    wandb.log({"Inventory": np.squeeze(info["inv"])}, commit=False)
                wandb.log(env.env_type, commit=False)
                wandb.log({
                    "Epoch": epoch,
                    "t": t,
                    "Action": info["action"],
                    "Reward": np.squeeze(info["reward"]),
                    "True_Reward": np.squeeze(info.get("true_reward", info["reward"])),
                    "Cumulative_Reward": cumulative_reward,
                    "Cumulative_True_Reward": cumulative_true_reward,
                }, commit=True)
        del(info_hist)
        
        # -------------------------------------------------- validation?
        if experiment_type == "train_val" and ((epoch + 1) % val_every == 0):
            assert eval_envs is not None, "`eval_envs` must be provided for train_val"
            cum_R, cum_True_R = evaluate_val(agent=agent,
                         envs=eval_envs,
                         epoch=epoch,
                         core_cls=core_cls,
                         use_wandb=use_wandb,
                         eval_step_info=eval_step_info)
            # -------------------------------------------------- early stopping
            if early_stopping_handler and early_stopping_handler.add_result(cum_R, cum_True_R):
                logging.info(f"Early‑stopping triggered at epoch {epoch+1}")
                break

        
        # -------------------------------------------------- prepare next epoch
        env.train()
        agent.train()
        gc.collect()
    # ==================================================================
    # 3) FINAL TEST‑SET EVALUATION  (optional)
    # ==================================================================
    # if eval_envs is not None:
    #     evaluate_test(agent=agent,
    #                   envs=eval_envs,
    #                   use_wandb=use_wandb,
    #                   epoch=epoch,
    #                   core_cls=core_cls,
    #                   eval_step_info=eval_step_info)
````

````python
#| export
def run_hp_experiment(agent: BaseAgent,
                train_envs: List = None,                 # envs used for training
                eval_envs: List = None,                  # envs used for val / test
                val_every: int = 10,    
                n_epochs: int= None,
                n_steps: int = None,
                n_steps_per_fit: int = 1,
                n_episodes_per_fit: int = None, 
                early_stopping_handler: Union[EarlyStoppingHandler, None] = None,
                save_best: bool = True,
                performance_criterion: str = "J",  # or "R"
                tracking: Union[str, None] = None,  # e.g., "wandb"
                results_dir: str = "results",
                run_id: Union[str, None] = None,
                print_freq: int = 1,
                eval_step_info=False):
    """
    Run an experiment with the given agent and environment for n_epochs.
    Automatically detects the training mode and runs accordingly.
    """
    epoch_results = []
    if run_id is None:
        run_id = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
    core_cls  = MetaCore if getattr(agent, "agent_name", "") == "RL2PPO" else Core
    experiment_dir = f"{results_dir}/{run_id}"
    print(f"Experiment directory: {experiment_dir}")
    logging.info("Starting experiment")

    for epoch in trange(n_epochs):
        env = train_envs[epoch % len(train_envs)]
        env.reset()
        core = core_cls(agent, env)

        if epoch > 0 and hasattr(agent, 'update_env'):
            agent.update_env(env)
        if hasattr(agent, "reset_hidden"):
            agent.reset_hidden()
        core.learn(
            n_steps=n_steps,
            n_steps_per_fit=n_steps_per_fit,
            n_episodes_per_fit=n_episodes_per_fit,
            quiet=False,
            n_episodes=1
        )

        # 2) score on training env
        info_history = env.get_info_history()
        cumulative_reward = 0.0
        cumulative_true_reward = 0.0
        for info in info_history.values():
            reward = np.squeeze(info["reward"])
            true_reward = np.squeeze(info.get("true_reward", reward))
            cumulative_reward += reward
            cumulative_true_reward += true_reward
        # drop your local ref
        del info_history
        # clear the env’s history dict in place
        env.get_info_history().clear()

        # 3) validation
        if (epoch + 1) % val_every == 0:
            assert eval_envs is not None, "`eval_envs` must be provided for train_val"
            if hasattr(agent, "reset_hidden"):
                agent.reset_hidden()
            agent.eval()
            all_Cum_True_R, all_Cum_R = [], []
            for venv in eval_envs:
                venv.reset()
                core_val = core_cls(agent, venv)

                core_val.evaluate(quiet=True, n_episodes=1)
            
                info_hist = venv.get_info_history()
                cum_R, cum_TR = 0.0, 0.0
                for _, info in info_hist.items():
                    cum_R += np.squeeze(info["reward"])
                    cum_TR += np.squeeze(info.get("true_reward", info["reward"]))
                all_Cum_R.append(cum_R)
                all_Cum_True_R.append(cum_TR)
                # cleanup this eval env
                del info_hist
                venv.get_info_history().clear()

            logging.info(f"[VAL] mean True_R={np.mean(all_Cum_True_R):.2f}  mean R={np.mean(all_Cum_R):.2f}")

            agent.train()
            val_metrics = {
                "val_reward": np.mean(all_Cum_R),
                "val_true_reward": np.mean(all_Cum_True_R),
            }
            epoch_results.append({"epoch": epoch, "val_metrics": val_metrics})

            if early_stopping_handler and early_stopping_handler.add_result(
                val_metrics["val_reward"], val_metrics["val_true_reward"]
            ):
                logging.info(f"Early‑stopping triggered at epoch {epoch+1}")
                # final cleanup before return
                env.get_info_history().clear()
                for v in eval_envs:
                    v.get_info_history().clear()
                gc.collect()
                return epoch_results

        # 4) end‑of‑epoch housekeeping
        env.train()
        agent.train()
        # clear any lingering history on all envs
        env.get_info_history().clear()
        for v in eval_envs:
            v.get_info_history().clear()


    return epoch_results
````

````python
#| export

def test_agent(agent: BaseAgent,
            env: BaseEnvironment,
            return_dataset = False,
            save_features = False,
            tracking = None, # other: "wandb",
            eval_step_info = False,
):

    """
    Tests the agent on the environment for a single episode
    """
    
    # TODO make it possible to save dataset via tracking tool

    # Run the test episode
    dataset = run_test_episode(env, agent, eval_step_info, save_features = save_features)

    # Calculate the score
    R, J = calculate_score(dataset, env)

    if tracking == "wandb":
        mode = env.mode
        wandb.log({f"{mode}/R": R, f"{mode}/J": J})

    if return_dataset:
        return R, J, dataset
    else:
        return R, J

def run_test_episode(   env: BaseEnvironment, # Any environment inheriting from BaseEnvironment
                        agent: BaseAgent, # Any agent inheriting from BaseAgent
                        eval_step_info: bool = False, # Print step info during evaluation
                        save_features: bool = False, # Save features (observation) of the dataset. Can be turned off since they sometimes become very large with many lag information

                ):

    """
    Runs an episode to test the agent's performance.
    It assumes, that agent and environment are initialized, in test/val mode
    and have done reset.
    """

    # Get initial observation
    obs = env.reset()

    dataset = []
    
    finished = False
    step = 0

    horizon = env.mdp_info.horizon
    
    while not finished:
        
        # Sample action from agent
        action = agent.draw_action(obs)

        # Take a step in the environment

        next_obs, reward, terminated, truncated, info = env.step(action)
        
        logging.debug("##### STEP: %d #####", env.index)
        logging.debug("reward: %s", reward)
        logging.debug("info: %s", info)
        logging.debug("next observation: %s", obs)
        logging.debug("truncated: %s", truncated)

        if save_features:
            sample = (obs, action, reward, next_obs, terminated, truncated) # unlike mushroom do not include policy_state
        else:
            sample = (None, action, reward, None, terminated, truncated)

        obs = next_obs
        
        dataset.append((sample, info))

        finished = terminated or truncated

        if eval_step_info:
            step += 1
            sys.stdout.write(f"\rStep {step}")
            sys.stdout.flush()

    if eval_step_info:
        print()

    return dataset

    
````

### Important notes on running experiments


**Training mode**:

* Agents have either a training mode ```direct_fit``` or ```epochs_fit``` or ```env_interaction```. ```direct_fit``` means that agents are called with a single call to the fit method, providing the full X and Y dataset. ```epochs_fit``` means that agents are training iteratively via epochs. It is assumed that they then have access to the dataloader.

**Train, val, test mode**:

* The function always sets the agent and environment to the approproate dataset mode (and thereofore indirectly the dataloader via then environment).

**Early stopping**:

* Can be optionally applied for ```epochs_fit``` and ```env_interaction``` agents.

**Save best agent**:

* The ```save_agent()``` functions, given the ```save_best```param is ```True```, will save the best agent based on the validation score.

* At test time at a later point, one can then load the best agent and evaluate it on the test set (not done automatically by this function).

**Logging**:

* By setting logging to ```"wandb"``` the function will log J and R to wandb.
